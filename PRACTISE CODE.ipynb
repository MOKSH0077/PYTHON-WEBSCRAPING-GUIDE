requests: A library for making HTTP requests (GET, POST, etc.). It is used to fetch web pages or APIs from the internet.

lxml: A library for parsing XML and HTML very fast. It is often used with BeautifulSoup for better performance.

bs4 (BeautifulSoup4): A library used for web scraping. It helps extract data from HTML and XML documents easily.


```python
import requests
```

```python
import bs4
```

```python
take_requests=requests.get('https://en.wikipedia.org/wiki/Magnus_Carlsen')
```

```python
code_version=bs4.BeautifulSoup(take_requests.text,'lxml')
```

code_version

```python
image_grab=code_version.select('.infobox-image')[0]
```

```python
image_grab
```

```python
requests_for_image_grab=requests.get('https://upload.wikimedia.org/wikipedia/commons/thumb/5/5f/MagnusCarlsen24.jpg/220px-MagnusCarlsen24.jpg')
```

```python
if requests_for_image_grab.status_code == 200:
    with open("carlsen.jpg", "wb") as f:
        f.write(requests_for_image_grab.content)
```

```python
print(requests_for_image_grab.headers.get("Content-Type"))
```
